# Lab 5 - Web Security & Wapiti

## Aim

- Use **Wapiti** as a black-box web vulnerability scanner.
- Run a scan against a deliberately vulnerable test application.
- Open and briefly interpret the resulting HTML report.

> Wapiti itself is run from the command line; the notebook focuses on the workflow
> and a helper function for opening the generated report.


### Example Wapiti Command (run in a terminal)

```bash
wapiti -u https://juice-shop.herokuapp.com -o juice_scan -f html

```

This will:

- Crawl the target application.
- Test parameters for issues such as XSS, SQL injection, etc.
- Save a report in the `gruyere_scan/` directory.


````python
import os
import glob
from IPython.display import display, HTML

# This must match the -o argument you gave to Wapiti
out_name = "juice_scan"

def find_html_report(out):
    # If "out" is an html file, use it directly
    if os.path.isfile(out) and out.lower().endswith(".html"):
        return os.path.abspath(out)

    # If "out" is a directory, look for index.html or the newest html file inside it
    if os.path.isdir(out):
        index_path = os.path.join(out, "index.html")
        if os.path.exists(index_path):
            return os.path.abspath(index_path)

        matches = glob.glob(os.path.join(out, "*.html"))
        if matches:
            matches.sort(key=os.path.getmtime, reverse=True)
            return os.path.abspath(matches[0])

    # Fallback: search in subfolders for any html file that looks like a report
    candidates = glob.glob(out + "/**/*.html", recursive=True) + glob.glob(out + "*.html")
    if candidates:
        candidates.sort(key=os.path.getmtime, reverse=True)
        return os.path.abspath(candidates[0])

    return None

report_path = find_html_report(out_name)

if report_path:
    print("Report file found:", report_path)
    try:
        with open(report_path, "r", encoding="utf-8") as f:
            html = f.read()
        display(HTML(html))
    except Exception as e:
        print("Could not render HTML inline. Error:", e)
        print("Open the file in your browser instead:", report_path)
else:
    print("No HTML report found. Check that the scan produced the report and that out_name matches the -o argument passed to Wapiti.")
````
